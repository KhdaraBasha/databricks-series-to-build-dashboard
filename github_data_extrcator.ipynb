### <strong>List of modules to perform data extraction</strong>

```python
# import modules
import pandas as pd
import requests
import json
import io
from datetime import datetime
import os
import configparser
from dotenv import load_dotenv
```

```python

# Load environment variables from .env file
load_dotenv('.config')  # Change to your .env file name if needed

# Access environment variables
baseurl = os.getenv('url')
```

### <strong>DECLARE LIST OF FILES</strong>

```python
# List of CSV Files
csv_files = [
    "Emissions_Data_2023.csv",
    "customerscsv.csv",
    "orderscsv.csv",
    "productscsv.csv"
]

# List of JSON Files
json_files = [
    "customersjson.json",
    "ordersjson.json",
    "productsjson.json"
]
```

### <strong>CREATE A FUNCTION TO EXTRACT CSV FILES</strong>

```python
# CSV Data Extrcation File
def csv_files_data_extrcat(file_name: str, baseurl: str):
    dataframes = {}
    url = baseurl + file_name
    response = requests.get(url)
    response.raise_for_status()
    print(f"Extracting data from {file_name}...")
    if response.status_code == 200:
        # Create a DataFrame from the CSV data
        df = pd.read_csv(io.StringIO(response.text))
        print(f"Data from {file_name} extracted successfully!")
        name = file_name.replace(".csv","")
        df.to_csv(f"extracted_data/{name}.csv", index=False)
        print(f"✅ Loaded {file_name} — {df.shape[0]} rows, {df.shape[1]} cols")
```

```python
# JSON Data Extrcation File
def json_files_data_extrcat(file_name: str, baseurl: str):
    dataframes = {}
    url = baseurl + file_name
    response = requests.get(url)
    response.raise_for_status()
    print(f"Extracting data from {file_name}...")
    if response.status_code == 200:
        # Parse JSON data
        json_data = response.json()
        # Create a DataFrame from the JSON data
        df = pd.DataFrame(json_data)
        print(f"Data from {file_name} extracted successfully!")
        name = file_name.replace(".json", "")
        df.to_json(f"extracted_data/{name}.json", orient='records', indent=2)
        print(f"✅ Loaded {file_name} — {df.shape[0]} rows, {df.shape[1]} cols")
        return df
```

```python
if __name__ == "__main__":
    # Create the 'extracted_data' directory if it doesn't exist
    if not os.path.exists('extracted_data'):
        os.makedirs('extracted_data')
    # Extract CSV files
    for file in csv_files:
        csv_files_data_extrcat(file, baseurl)
    # Extract JSON files
    for file in json_files:
        json_files_data_extrcat(file, baseurl)
```
